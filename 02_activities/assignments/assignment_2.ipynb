**Problem 1 Starter Code**

You may assume that `root` is a binary tree that has been built, i.e., you can access a node/leaf by calling `root.left.right.left.right...`, so you don't have to provide a tree-building function to convert an input list to a binary tree as an object of the `TreeNode` class.

```python
# Definition for a binary tree node.
class TreeNode(object):
    def __init__(self, val = 0, left = None, right = None):
        self.val = val
        self.left = left
        self.right = right

from collections import deque

#Tree building function
def root_to_tree(arr):
    if not arr:
        return None
    
    root = TreeNode(arr[0]) # Create the root node from the first element
    queue = deque([root])
    i = 1
      
    while queue and i < len(arr): # Process the list level by level
        curr = queue.popleft()
        
        if i < len(arr) and arr[i] is not None: # Left branch
            curr.left = TreeNode(arr[i])
            queue.append(curr.left)
        i += 1
       
        if i < len(arr) and arr[i] is not None:  # Right branch
            curr.right = TreeNode(arr[i])
            queue.append(curr.right)
        i += 1
        
    return root

def is_duplicate(root: TreeNode) -> int:
  # Code your solution here #
  dup = set() #Blank set to store elements in order to check for duplicates
  queue = deque([root]) #deque queue for storing
 
  while queue:
     tree = queue.popleft()

     if tree.val in dup:
        return tree.val
     
     dup.add(tree.val)
     
     if tree.left:
        queue.append(tree.left)
     if tree.right:
        queue.append(tree.right)
        
  return -1

root = [10, 9, 8, 7]

tree_root = root_to_tree(root)

is_duplicate(tree_root)
```


4. Copy the solution your partner wrote. 


```python
# Your answer here
from typing import List

def move_zeros_to_end(nums: List[int]) -> List[int]:
    # TODO
    result = []
    zero_count = 0

    # Go through each number in the list
    for num in nums:
        if num == 0:
            zero_count += 1
        else:
            result.append(num)

    # Add all zeros at the end
    for i in range(zero_count):
        result.append(0)

    return result
```

5. Explain why their solution works in your own words.


```python
# Your answer here
First, the code makes a blank list named result and variable zero_count which starts as 0. Then it uses for loop to check each element in the list if it is a zero.
If the element is zero, the zero_count increases by 1. If not, the element is added to result.
Once the first loop is complete, the next for loop adds zeros to result based on zero_count.
Since the code goes through the list element-by-element, the relative order of non-zero integer is maintained,
while zeros are added based on the number of zeros (zero_count), it returns the expected output.
```
